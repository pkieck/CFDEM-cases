```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
import csv
%matplotlib inline
```

```python
def read_recurrence_matrix(filename):
    recurrence_matrix = []
    with open(filename, 'r') as f:
        reader = csv.reader(f)
        
        current_row = []
        for row in reader:
            if '(' in row:
                current_row = []
                continue
            if ')' in row:
                recurrence_matrix.append(current_row)
                continue
            try:
                current_row.append(float(row[0]))
            except ValueError:
                continue
    return 1.-np.array(recurrence_matrix)

def read_recurrence_path(filename):
    path = []
    with open(filename, 'r') as f:
        reader = csv.reader(f)
        for row in reader:
            try:
                new_strings = row[0].replace('(','').replace(')','').split(' ')
                new_entry   = [int(x) for x in new_strings]
                if len(new_entry) == 2:
                    path.append(new_entry) 
            except:
                continue
    return np.array(path,dtype=int)

def plot_recurrence_path(recurrence_path):
    t = 0
    for j in recurrence_path:
        dt = j[1]-j[0]
        plt.plot(np.array([t,t+dt]),j, color='black')
        t += dt
    plt.ylabel('$t_{Rec}$')
    plt.xlabel('$t$')
```

```python
b = read_recurrence_path('recurrencePath')
plot_recurrence_path(b)
```

```python
c = read_recurrence_matrix('recurrenceMatrix')
plt.imshow(c, origin='lower', cmap=mpl.cm.get_cmap('inferno'))
plt.xlabel('$t$')
plt.ylabel('$t$')
plt.colorbar()
```
