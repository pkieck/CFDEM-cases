```python
import numpy as np
import matplotlib.pyplot as plt
import os
import csv
```

```python
def extract_vector(line):
    if '(' not in line or ')' not in line:
        return
    result_vector = []
    started = False
    tmp_letter = ""
    for letter in line:
        if '(' in letter:
            started = True
            continue
        if not started:
            continue
        if ' ' in letter or ')' in letter:
            result_vector.append(float(tmp_letter))
            tmp_letter = ""
        if ')' in letter:
            break
        tmp_letter += letter
    return result_vector
        
def get_positions(time):
    path = str(time) + '/lagrangian/recurrenceCloud/positions'
    result_positions = []
    with open(path,'r') as f:
        for row in f:
            addition = extract_vector(row.strip())
            if addition:
                result_positions.append(addition)
    return result_positions

def get_ids(time):
    path = str(time) + '/lagrangian/recurrenceCloud/origId'
    result_ids = []
    with open(path, 'r') as f:
        started = False
        for row in f:
            if '(' in row:
                started = True
                continue
            if not started:
                continue
            if ')' in row:
                break
            try:
                result_ids.append(int(row.strip()))
            except:
                continue
    return result_ids
            
def get_procs(time):
    path = str(time) + '/lagrangian/recurrenceCloud/origProcId'
    result_procs = []
    with open(path, 'r') as f:
        started = False
        for row in f:
            if '(' in row:
                started = True
                continue
            if not started:
                continue
            if ')' in row:
                break
            try:
                result_procs.append(int(row.strip()))
            except:
                continue
    return result_procs

def get_offsets(time):
    proc_ids = np.array(get_procs(time))
    procs = np.unique(proc_ids)
    ids = np.array(get_ids(time))
    offsets = [ 0 ]
    for i, proc in enumerate(np.unique(proc_ids)):
        subset = proc_ids==proc
        ids[subset] += offsets[-1]+1
        offsets.append(max(ids[subset]))
    assert(max(ids)==len(ids))
    return offsets

def renumber_ids(id_list, proc_list, offset_set):
    proc_ids = np.array(proc_list)
    procs = np.unique(proc_ids)
    ids = np.array(id_list)
    offsets = offset_set[:]
    for i, proc in enumerate(np.unique(proc_ids)):
        subset = proc_ids==proc
        ids[subset] += offsets[i]
    return ids

def get_renumbered_ids(time, offsets):
    ids = get_ids(time)
    procs = get_procs(time)
    return renumber_ids(ids, procs, offsets)

  
```
